```sql
-- Range will treat duplicates as same row
```

```sql
SELECT 
    *,
    SUM(amount) OVER (
        PARTITION BY department
        ORDER BY amount  ROWS BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW
    ) AS running_total
FROM workspace.practice.order_window;
```

```sql
SELECT 
    *,
    SUM(amount) OVER (
        PARTITION BY department
        ORDER BY amount  RANGE BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW
    ) AS running_total
FROM workspace.practice.order_window;
```

```sql
SELECT 
    *,
    SUM(amount) OVER (
        PARTITION BY department
        ORDER BY amount  ROWS BETWEEN 2 PRECEDING AND CURRENT ROW
    ) AS running_total
FROM workspace.practice.order_window;
```

```sql
SELECT 
    *,
    SUM(amount) OVER (
        PARTITION BY department
        ORDER BY amount   ROWS BETWEEN 1 PRECEDING AND 1 FOLLOWING
    ) AS running_total
FROM workspace.practice.order_window;
```
